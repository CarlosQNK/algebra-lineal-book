# Operaciones elementales de fila de una matriz (Eliminación gaussiana)

El proceso de eliminación de Gauss-Jordan es parte del álgebra lineal computacional

Una **operación elemental de fila (o columna)** transforma una matriz en otra matriz equivalente mediante una de las siguientes operaciones:

1. **Intercambio de dos filas (o columnas):**
   $$
   \begin{pmatrix}
   \color{red}{1} & \color{red}{2} & \color{red}{-3} \\
   \color{blue}{3} & \color{blue}{0} & \color{blue}{1}
   \end{pmatrix}
   \xrightarrow{F_1\leftrightarrow F_2}
   \begin{pmatrix}
   \color{blue}{3} & \color{blue}{0} & \color{blue}{1} \\
   \color{red}{1} & \color{red}{2} & \color{red}{-3}
   \end{pmatrix}
   $$

   La fila roja pasa a la segunda posición y la fila azul pasa a la primera. Esto representa:
   $$
   A=\begin{pmatrix}
   \color{red}{1} & \color{red}{2} & \color{red}{-3} \\
   \color{blue}{3} & \color{blue}{0} & \color{blue}{1}
   \end{pmatrix},
   \qquad
   B=\begin{pmatrix}
   \color{blue}{3} & \color{blue}{0} & \color{blue}{1} \\
   \color{red}{1} & \color{red}{2} & \color{red}{-3}
   \end{pmatrix}
   $$

   **Intercambio entre la primera y la tercera columna:**
   $$
   \begin{pmatrix}
   \color{red}{1} & 2 & \color{blue}{-3} \\
   \color{red}{3} & 0 & \color{blue}{1}
   \end{pmatrix}
   \xrightarrow{C_1\leftrightarrow C_3}
   \begin{pmatrix}
   \color{blue}{-3} & 2 & \color{red}{1} \\
   \color{blue}{1} & 0 & \color{red}{3}
   \end{pmatrix}
   $$

   La columna roja pasa a la tercera posición y la columna azul pasa a la primera.

2. **Multiplicación de una fila por un número diferente de cero:**

   $$
   \begin{pmatrix}
   \color{red}{1} & \color{red}{2} & \color{red}{-3} \\
   3 & 0 & 1
   \end{pmatrix}
   \xrightarrow{F_1\leftarrow 2F_1}
   \begin{pmatrix}
   \color{red}{2} & \color{red}{4} & \color{red}{-6} \\
   3 & 0 & 1
   \end{pmatrix}
   $$

   Se multiplica toda la primera fila por el escalar no nulo \(2\).

   **Multiplicación de una columna por un número diferente de cero:**

   $$
   \begin{pmatrix}
   1 & \color{blue}{2} & -3 \\
   3 & \color{blue}{0} & 1
   \end{pmatrix}
   \xrightarrow{C_2\leftarrow (-1)C_2}
   \begin{pmatrix}
   1 & \color{blue}{-2} & -3 \\
   3 & \color{blue}{0} & 1
   \end{pmatrix}
   $$

   Se multiplica toda la segunda columna por el escalar no nulo \(-1\).

2. **Multiplicación de una fila por un escalar no nulo:**
   \[
   F_i \longleftarrow cF_i, \qquad c \neq 0
   \]

3. **Reemplazo de una fila por ella misma más un múltiplo de otra fila:**
   \[
   F_i \longleftarrow F_i + cF_j, \qquad i \neq j
   \]

Estas operaciones preservan la equivalencia por filas de la matriz y se utilizan, por ejemplo, para obtener su forma escalonada mediante el método de Gauss.
